## 1. De la entrada al dato útil

`input()` muestra una pregunta, espera texto escrito por la persona y devuelve una cadena (`str`). Aunque se escriba `18.4`, Python recibe inicialmente el texto `"18.4"`. Para calcular se convierte con `float()`; para conteos enteros, con `int()`. La conversión falla si el texto no representa un número válido.

```python
# Simulamos lo que escribiría una persona en una estación meteorológica.
lluvia_texto = "14.8"
lluvia_mm = float(lluvia_texto)

print(type(lluvia_texto))
print(type(lluvia_mm))
print(f"Precipitación registrada: {lluvia_mm:.1f} mm")
```

**Lectura:** `lluvia_texto` es `str`; `float()` transforma el texto a número decimal. La f-string inserta el valor y `:.1f` muestra un decimal. La unidad `mm` hace interpretable el dato.

Para probar teclado, cambia la asignación simulada por `lluvia_texto = input("Lluvia registrada (mm): ")` y ejecuta esta celda. No ejecutes todo el cuaderno con entradas interactivas activas.

## 2. Contexto meteorológico: lluvia acumulada

Una persona monitora la precipitación de una estación y estima el acumulado del día a partir de la lluvia observada y una lectura complementaria.

**Entrada:** precipitación observada y adicional, en milímetros. **Proceso:** sumar ambas mediciones. **Salida:** mostrar el total con dos decimales.

```python
# Sustituye los valores simulados por input() para una prueba interactiva.
estacion = "La Argelia"
lluvia_observada_mm = float("14.8")
lluvia_adicional_mm = float("5.2")
# estacion = input("Nombre de la estación: ")
# lluvia_observada_mm = float(input("Lluvia observada (mm): "))
# lluvia_adicional_mm = float(input("Lluvia adicional (mm): "))

lluvia_total_mm = lluvia_observada_mm + lluvia_adicional_mm
print(f"Estación: {estacion}")
print(f"Precipitación acumulada: {lluvia_total_mm:.2f} mm")
```

## 3. Contexto de calidad del agua: turbidez

En una ficha de muestreo se registra el sitio y la turbidez medida en unidades nefelométricas (NTU). Aquí nos enfocamos en leer y reportar la medición, sin clasificarla.

```python
sitio_muestreo = "Río Malacatos - puente de Vilcabamba"
turbidez_texto = "8.76"
# En la prueba interactiva, reemplaza la línea anterior por:
# turbidez_texto = input("Turbidez medida (NTU): ")
turbidez_ntu = float(turbidez_texto)

print(f"Sitio: {sitio_muestreo}")
print(f"Turbidez registrada: {turbidez_ntu:.2f} NTU")
```

**Decisión de tipo:** turbidez puede incluir decimales, por eso usamos `float()`. El nombre del sitio permanece como texto. La medición se reporta con dos decimales, sin afirmar por sí sola si el agua cumple una norma.

## 4. Contexto de biodiversidad: conteo de individuos

Durante un muestreo se registra una especie observada y el número de individuos. Un conteo es entero; no tendría sentido ingresar `3.5` individuos.

```python
parcela = "Bosque protector El Colambo"
especie = "Alnus acuminata"
individuos_texto = "23"
# Para probar input(): individuos_texto = input("Individuos observados: ")
numero_individuos = int(individuos_texto)

print(f"Parcela: {parcela}")
print(f"Especie: {especie}")
print(f"Individuos observados: {numero_individuos}")
```

## 5. Contexto de gestión de residuos: masa recolectada

Un equipo registra la masa de residuos recuperables recolectados en una jornada y calcula cuánto se reunió entre dos sectores.

```python
sector = "Quebrada San Cayetano"
masa_sector_a_kg = float("18.5")
masa_sector_b_kg = float("11.25")
# Alternativa interactiva:
# masa_sector_a_kg = float(input("Masa del sector A (kg): "))
# masa_sector_b_kg = float(input("Masa del sector B (kg): "))
masa_total_kg = masa_sector_a_kg + masa_sector_b_kg

print(f"Jornada de limpieza: {sector}")
print(f"Masa recuperada: {masa_total_kg:.2f} kg")
```

## 6. Errores frecuentes al ingresar datos

- **Olvidar la conversión:** `input()` devuelve texto; sumar dos entradas de texto puede concatenarlas en vez de sumar cantidades.
- **Usar el separador decimal equivocado:** escribe `7.25`, no `7,25`.
- **Convertir un texto descriptivo:** `float("Río Zamora")` no es válido. Los nombres se conservan como `str`.
- **Elegir el tipo incorrecto:** mediciones con decimales usan `float()`; conteos enteros usan `int()`.

Todavía no usamos manejo avanzado de errores. En un programa real se debe revisar el dato recibido antes de usarlo.
