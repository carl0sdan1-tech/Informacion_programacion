## 1. Problema, algoritmo y programa

- **Problema:** situación concreta que requiere una solución.
- **Algoritmo:** secuencia finita, ordenada y comprensible de pasos para resolver el problema. Puede escribirse en lenguaje cotidiano, pseudocódigo o un diagrama.
- **Programa:** algoritmo expresado en un lenguaje de programación que una computadora puede ejecutar.

**Ejemplo ambiental:** queremos conocer cuántos días de un mes registraron lluvia superior a un umbral. El problema es la pregunta; el algoritmo explica cómo contar; el programa será su implementación posterior en Python.

## 2. Pensamiento computacional

**Descomposición** consiste en dividir una tarea compleja en partes pequeñas. Para resumir lluvias mensuales podemos: (1) identificar las observaciones disponibles, (2) decidir el umbral, (3) contar las que lo superan y (4) comunicar el conteo.

**Abstracción** conserva lo necesario para responder la pregunta y deja de lado lo irrelevante. Para contar días lluviosos por encima del umbral necesitamos los valores diarios y el umbral; el color del cuaderno de campo no afecta el conteo.

**Reconocimiento de patrones** identifica operaciones que se repiten. En cada día comparamos la lluvia con el mismo umbral y actualizamos un conteo.

## 3. Entrada, proceso y salida

Una manera de organizar un algoritmo es **Entrada → Proceso → Salida (E-P-S)**.

| Etapa | Pregunta | Ejemplo: celdas con bosque > 50 % |
|---|---|---|
| Entrada | ¿Qué datos hacen falta? | Porcentajes de bosque de las celdas y total de celdas |
| Proceso | ¿Qué operaciones resuelven la pregunta? | Contar las celdas con valor > 50 y dividir por el total; convertir la proporción a porcentaje |
| Salida | ¿Qué debe recibir quien usa el resultado? | Número de celdas, total evaluado y porcentaje, con su unidad |

La salida debe responder al problema; no basta con imprimir un número sin explicar qué representa.

## 4. Pseudocódigo: describir sin depender de Python

El pseudocódigo es una forma sencilla de escribir los pasos que debe seguir una computadora para resolver un problema, utilizando palabras comunes y una estructura ordenada.

Inicio
    lluvias <- [5, 12, 8, 15, 10]
    umbral <- 10
    contador <- 0

    Para cada lluvia en lluvias:
        Si lluvia > umbral:
            contador <- contador 0

    Mostrar contador
Fin

`<-` representa asignación: guardar o actualizar un valor. La comparación `>` pregunta si un dato supera estrictamente el umbral; un valor igual al umbral no cuenta.

## 4. Diagrama de flujo

Un flujograma representa los pasos y el orden de ejecución con símbolos convencionales.

| Símbolo habitual | Significado |
|---|---|
| Óvalo | Inicio o fin |
| Paralelogramo | Entrada o salida de datos |
| Rectángulo | Proceso o asignación |
| Rombo | Decisión con rutas según la respuesta |
| Flecha | Dirección del flujo |


             (Inicio)
                ↓
     [Ingresar datos de lluvia]
                ↓
       [Establecer umbral]
                ↓
         [contador = 0]
                ↓
       [Leer una lluvia]
                ↓
       ◇ ¿Lluvia > umbral? ◇
          /           \
        Sí             No
        ↓               |
[contador = contador + 1]
        |               |
        └───────┬───────┘
                ↓
     ◇ ¿Quedan días por revisar? ◇
          /           \
        Sí             No
        |               ↓
        |       [Mostrar contador]
        |               ↓
        |             (Fin)
        |
        └──→ [Leer siguiente lluvia]

La flecha de retorno del ciclo debe volver a revisar la siguiente observación hasta terminar la lista. En un diagrama dibujado, usa conectores para que las rutas no queden ambiguas.

## 5. Prueba de escritorio

La **prueba de escritorio** consiste en revisar manualmente cómo funciona un algoritmo, utilizando datos sencillos, sin necesidad de ejecutarlo en una computadora. Permite comprobar que las instrucciones sean correctas y detectar posibles errores antes de programar.

**Ejemplo:** queremos contar cuántos días la lluvia superó los 10 mm. Para ello, tenemos los siguientes datos:

- **Lluvias registradas:** `[0, 12, 8, 25]` mm.
- **Umbral:** 10 mm.
- **Contador inicial:** 0.

El algoritmo revisa cada valor de lluvia. Si es mayor que 10 mm, aumenta el contador en una unidad. En caso contrario, el contador no cambia.

### Tabla de prueba de escritorio

| Día | Lluvia (mm) | ¿Supera 10 mm? | Contador |
|:---:|---:|:---:|---:|
| Inicio | — | — | 0 |
| 1 | 0 | No | 0 |
| 2 | 12 | Sí | 1 |
| 3 | 8 | No | 1 |
| 4 | 25 | Sí | 2 |

### ¿Cómo interpretamos la tabla?

1. **Inicio:** el contador comienza en cero porque todavía no hemos revisado ningún día.
2. **Día 1:** se registran 0 mm. Como no supera los 10 mm, el contador permanece en 0.
3. **Día 2:** se registran 12 mm. Como supera los 10 mm, sumamos uno al contador. Ahora vale 1.
4. **Día 3:** se registran 8 mm. Como no supera los 10 mm, el contador mantiene su valor de 1.
5. **Día 4:** se registran 25 mm. Como supera los 10 mm, sumamos uno al contador. Su valor final es 2.

### Resultado

De los **4 días evaluados**, solamente **2 días** registraron lluvias superiores a 10 mm.

**Importante:** la condición `lluvia > 10` significa que solamente contamos los valores mayores que 10. Si un día registra exactamente 10 mm, no se cuenta. Para incluirlo, debemos utilizar `lluvia >= 10` (mayor o igual que 10).

**En conclusión:** la prueba de escritorio permite observar cómo cambian las variables, comprobar las condiciones y verificar que el algoritmo produce el resultado esperado antes de escribirlo en Python.

## 6. Segundo contexto: cobertura de bosque nativo

La **cobertura de bosque nativo** indica qué porcentaje de una superficie está ocupado por bosque natural.

**Problema:** queremos conocer qué porcentaje de las celdas evaluadas tiene una cobertura de bosque nativo superior al 50 %.

### Ejemplo

Supongamos que tenemos **5 celdas**, cada una con un porcentaje diferente de cobertura de bosque nativo:

`[65, 40, 52, 50, 80]`

Esto significa que:
- **Celda 1:** tiene 65 % de cobertura de bosque.
- **Celda 2:** tiene 40 % de cobertura de bosque.
- **Celda 3:** tiene 52 % de cobertura de bosque.
- **Celda 4:** tiene 50 % de cobertura de bosque.
- **Celda 5:** tiene 80 % de cobertura de bosque.

Queremos contar cuántas celdas **superan el 50 % de cobertura**.

### Paso 1. Revisar cada celda

| Celda | Cobertura de bosque (%) | ¿Supera el 50 %? | Contador |
|:---:|---:|:---:|---:|
| Inicio | — | — | 0 |
| 1 | 65 | Sí | 1 |
| 2 | 40 | No | 1 |
| 3 | 52 | Sí | 2 |
| 4 | 50 | No | 2 |
| 5 | 80 | Sí | 3 |

**Interpretación:** las celdas 1, 3 y 5 tienen una cobertura superior al 50 %. Por tanto, **3 de las 5 celdas cumplen la condición**.

La celda 4 tiene exactamente 50 %, por lo que no se cuenta, porque estamos buscando valores mayores que 50 (`> 50`).

### Paso 2. Calcular el porcentaje de celdas que cumplen la condición

Utilizamos la siguiente fórmula:

$$
\text{Porcentaje}=\frac{\text{Celdas que cumplen la condición}}{\text{Total de celdas}}\times 100
$$

Reemplazamos los valores:

$$
\text{Porcentaje}=\frac{3}{5}\times 100=60\%
$$

### Paso 3. Interpretar el resultado

De las **5 celdas evaluadas**, **3 tienen más del 50 % de cobertura de bosque nativo**. Esto representa el **60 % del total de celdas**.

**Importante:** no debemos confundir los dos porcentajes:

- **50 %:** es el umbral de cobertura de bosque que utilizamos para evaluar cada celda.
- **60 %:** es el porcentaje de celdas que superaron ese umbral.

**En conclusión:** este ejemplo muestra cómo podemos utilizar un contador, una comparación y una operación matemática para resolver un problema ambiental mediante el pensamiento computacional.



## 7. Tercer contexto: identificar días secos y comunicar una posible alerta de sequía

Un **día seco**, para este ejercicio, es aquel en el que la precipitación registrada es menor a 1 mm.

**Problema:** queremos conocer cuántos días registraron menos de 1 mm de lluvia durante un período de observación.

### Ejemplo

Supongamos que tenemos los registros de precipitación de **6 días**:

`[0, 5, 0.5, 2, 0.8, 1]` mm

Queremos contar cuántos días tuvieron una precipitación **menor a 1 mm**.

### Paso 1. Identificar las entradas, el proceso y la salida

- **Entrada:** los valores de precipitación diaria `[0, 5, 0.5, 2, 0.8, 1]` mm y el umbral de 1 mm.
- **Proceso:** revisar cada día y aumentar el contador cuando la precipitación sea menor a 1 mm.
- **Salida:** número total de días secos identificados.

### Paso 2. Realizar la prueba de escritorio

Primero, establecemos `contador <- 0`.

| Día | Precipitación (mm) | ¿Es menor a 1 mm? | Operación | Contador |
|:---:|---:|:---:|:---:|---:|
| Inicio | — | — | Inicializar | 0 |
| 1 | 0 | Sí | 0 + 1 | 1 |
| 2 | 5 | No | No suma | 1 |
| 3 | 0.5 | Sí | 1 + 1 | 2 |
| 4 | 2 | No | No suma | 2 |
| 5 | 0.8 | Sí | 2 + 1 | 3 |
| 6 | 1 | No | No suma | 3 |

### Paso 3. Interpretar la tabla

1. **Día 1:** la precipitación es 0 mm. Como es menor a 1 mm, contamos el primer día seco.
2. **Día 2:** se registran 5 mm. No cumple la condición, por lo que el contador permanece en 1.
3. **Día 3:** se registran 0.5 mm. Cumple la condición y el contador aumenta a 2.
4. **Día 4:** se registran 2 mm. No cumple la condición y el contador permanece en 2.
5. **Día 5:** se registran 0.8 mm. Cumple la condición y el contador aumenta a 3.
6. **Día 6:** se registra exactamente 1 mm. No se cuenta porque la condición exige valores menores a 1 mm.

### Paso 4. Escribir el pseudocódigo

```text
Inicio
    precipitaciones <- [0, 5, 0.5, 2, 0.8, 1]
    contador <- 0

    Para cada lluvia en precipitaciones:
        Si lluvia < 1:
            contador <- contador + 1

    Mostrar contador
Fin
```

### Paso 5. Interpretar el resultado

De los **6 días evaluados**, **3 días** tuvieron precipitaciones inferiores a 1 mm.

Estos corresponden a los valores **0, 0.5 y 0.8 mm**.

**Importante:** debemos diferenciar las siguientes condiciones:

- `lluvia < 1`: cuenta únicamente los valores menores a 1 mm.
- `lluvia <= 1`: cuenta también los valores iguales a 1 mm.

Con la primera condición obtenemos **3 días secos**; con la segunda, **4 días**.

### Paso 6. Comunicar el resultado

Podemos presentar el resultado de la siguiente manera:

**Reporte:** "Durante los 6 días evaluados, se identificaron 3 días con precipitación inferior a 1 mm, equivalentes al 50 % del período analizado".

**En conclusión:** este ejercicio demuestra cómo utilizar condiciones y contadores para identificar días con precipitaciones muy bajas. Sin embargo, contar días secos no es suficiente para declarar una sequía. Para establecer una alerta real se necesita analizar un período más amplio y utilizar indicadores meteorológicos apropiados.


````python

# ======================================================
# ACTIVIDAD: ANÁLISIS DE COBERTURA DE BOSQUE NATIVO
# ======================================================

# OBJETIVO:
# Calcular cuántas celdas tienen una cobertura de
# bosque nativo superior al 50 % y determinar
# qué porcentaje representan del total.

# ------------------------------------------------------
# PASO 1. INGRESAR LOS DATOS
# ------------------------------------------------------

# Cada valor representa la cobertura de bosque
# nativo (%) de una celda.

cobertura = [65, 40, 52, 50, 80, 30, 75, 45, 90, 55]

# Valor que utilizaremos para comparar
umbral = 50

# Mostrar los datos
print("ANÁLISIS DE COBERTURA DE BOSQUE NATIVO")
print("--------------------------------------")
print("Coberturas registradas:", cobertura)
print("Umbral de comparación:", umbral, "%")


# ------------------------------------------------------
# PASO 2. CONTAR LAS CELDAS
# ------------------------------------------------------

# len() permite conocer cuántos datos tiene la lista.

total = len(cobertura)

print("\nTotal de celdas evaluadas:", total)


# ------------------------------------------------------
# PASO 3. COMPARAR CADA CELDA CON EL UMBRAL
# ------------------------------------------------------

# El símbolo > significa "mayor que".
# Python muestra True si cumple la condición
# y False si no la cumple.

print("\nCOMPARACIÓN DE COBERTURAS")

print("Celda 1:", cobertura[0], "% ->", cobertura[0] > umbral)
print("Celda 2:", cobertura[1], "% ->", cobertura[1] > umbral)
print("Celda 3:", cobertura[2], "% ->", cobertura[2] > umbral)
print("Celda 4:", cobertura[3], "% ->", cobertura[3] > umbral)
print("Celda 5:", cobertura[4], "% ->", cobertura[4] > umbral)
print("Celda 6:", cobertura[5], "% ->", cobertura[5] > umbral)
print("Celda 7:", cobertura[6], "% ->", cobertura[6] > umbral)
print("Celda 8:", cobertura[7], "% ->", cobertura[7] > umbral)
print("Celda 9:", cobertura[8], "% ->", cobertura[8] > umbral)
print("Celda 10:", cobertura[9], "% ->", cobertura[9] > umbral)


# ------------------------------------------------------
# PASO 4. CONTAR LAS CELDAS QUE CUMPLEN
# ------------------------------------------------------

# En Python:
# True representa 1 cuando realizamos una suma.
# False representa 0 cuando realizamos una suma.
#
# Por tanto, podemos sumar las comparaciones.

contador = sum([
    cobertura[0] > umbral,
    cobertura[1] > umbral,
    cobertura[2] > umbral,
    cobertura[3] > umbral,
    cobertura[4] > umbral,
    cobertura[5] > umbral,
    cobertura[6] > umbral,
    cobertura[7] > umbral,
    cobertura[8] > umbral,
    cobertura[9] > umbral
])

print("\nCeldas que superan el umbral:", contador)


# ------------------------------------------------------
# PASO 5. CALCULAR EL PORCENTAJE
# ------------------------------------------------------

# Fórmula:
# Porcentaje = (Celdas que cumplen / Total) * 100

porcentaje = (contador / total) * 100

print("Porcentaje de celdas:", porcentaje, "%")


# ------------------------------------------------------
# PASO 6. CALCULAR LA COBERTURA PROMEDIO
# ------------------------------------------------------

# sum() suma todos los porcentajes de cobertura.
# Dividimos esa suma para el número de celdas.

suma = sum(cobertura)

promedio = suma / total

print("Suma de coberturas:", suma)
print("Cobertura promedio:", promedio, "%")


# ------------------------------------------------------
# PASO 7. MOSTRAR EL RESUMEN FINAL
# ------------------------------------------------------

print("\n======================================")
print("           RESUMEN FINAL")
print("======================================")

print("Total de celdas evaluadas:", total)
print("Umbral utilizado:", umbral, "%")
print("Celdas que superan el umbral:", contador)
print("Porcentaje de celdas:", porcentaje, "%")
print("Cobertura promedio:", round(promedio, 2), "%")

print("======================================")
print("FIN DEL ANÁLISIS")
````
